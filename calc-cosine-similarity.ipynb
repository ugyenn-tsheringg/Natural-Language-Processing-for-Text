# Cosine Similarity
- Cosine similarity is a metric used to measure how similar two documents are, irrespective of their size. 
- Mathematically, it measures the cosine of the angle between two vectors projected in a multi-dimensional space.

# Common Use Cases

1. **Document Similarity and Clustering**: In content-based filtering systems, such as news aggregators or document management systems, cosine similarity can be used to suggest or cluster similar articles or documents. For instance, if a user reads an article about machine learning, the system can recommend other articles that have a high cosine similarity score with the read article.

2. **Recommendation Systems**: Cosine similarity is widely used in recommendation systems. For example, in a movie recommendation system, the plot summaries of two movies could be compared for similarity. If two movies have very similar plot summaries, it would make sense to recommend one to someone who has already expressed interest in the other.

3. **Search Engines**: Search engines use cosine similarity to rank documents according to their relevance to a given search query. The search engine calculates the cosine similarity between the search query and each document in its database, then it returns the documents with the highest scores.

4. **Semantic Analysis**: Companies often use cosine similarity in semantic analysis to understand customer sentiment. For example, by comparing customer reviews to a set of predefined positive or negative sentiments, businesses can get a better understanding of how their customers feel about their products or services.

5. **Resume Matching**: In HR and recruitment, cosine similarity can be used to match candidate resumes to job descriptions. The algorithm can calculate the cosine similarity between the job description and the resumes to find the most suitable candidates.

6. **Plagiarism Detection**: In academia or journalism, cosine similarity can be used to detect plagiarism. By comparing a document against a database of known documents, a high similarity score could indicate potential plagiarism.

# Import Package

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

# Create Function

```python
def calculate_cosine_similarity(text1, text2):
    vectorizer = CountVectorizer().fit_transform([text1, text2])
    print(vectorizer)
    vectors = vectorizer.toarray()
    print(vectors)
    cosine_sim = cosine_similarity(vectors)
    return cosine_sim[0][1]
```

# Calculate Similarity

```python
text1 = 'This is white cat.'
text2 = 'This is blue cat.'

print(calculate_cosine_similarity(text1, text2))
```

```python
text1 = 'This is white cat.'
text2 = 'This is white dog.'

print(calculate_cosine_similarity(text1, text2))
```
